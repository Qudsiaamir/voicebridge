```python
import cupy as cp

# Create a NumPy array
numpy_array = cp.array([1, 2, 3, 4, 5])

# Transfer the NumPy array to a CuPy array on the GPU
cupy_array = cp.asarray(numpy_array)

# Perform some operations on the GPU
result = cupy_array * 2

# Transfer the result back to a NumPy array (if needed)
result_numpy = cp.asnumpy(result)

# Print the results
print("NumPy Array:")
print(numpy_array)

print("\nCuPy Array:")
print(cupy_array)

print("\nResult on GPU:")
print(result)

print("\nResult transferred back to NumPy:")
print(result_numpy)
```

```python
import torch

# Check if GPU (CUDA) is available
if torch.cuda.is_available():
    # Get the number of available GPUs
    num_gpus = torch.cuda.device_count()
    print(f"Found {num_gpus} GPU(s).")

    # Print information about each GPU
    for i in range(num_gpus):
        gpu_name = torch.cuda.get_device_name(i)
        print(f"GPU {i}: {gpu_name}")
else:
    print("CUDA is not available. Running on CPU.")

    
```
